```julia
using OrdinaryDiffEq
using CairoMakie

f(u, p, t) = -u

prob = ODEProblem(f, 1.0, (0.0, 5.0))
sol = solve(prob, Tsit5())

fig = Figure()
ax = Axis(fig[1, 1], xlabel="Tiempo", ylabel="Solución")
lines!(ax, sol.t, sol.u)

display(fig)
```
